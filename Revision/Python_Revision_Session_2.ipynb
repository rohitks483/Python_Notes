# Part 1 — Lists
**A list is an ordered, changeable (mutable) collection**, written with `[ ]`. It can hold any types, and duplicates are allowed.

## 1.1 Indexing, slicing, replication
- Index starts at **0**. Negative index counts from the end (`-1` = last).
- Slice `lst[start:stop:step]` → `stop` is **not included**.
- 🆕 **Replication** with `*` repeats the list.

````python
fruits = ["apple", "banana", "cherry", "date", "elderberry"]

print(fruits[0], fruits[-1])    # apple elderberry
print(fruits[1:4])              # ['banana', 'cherry', 'date']
print(fruits[::2])              # every 2nd item
print(fruits[::-1])             # reversed

# 🆕 Replication
print([0] * 5)                  # [0, 0, 0, 0, 0]
print(["hi", "bye"] * 2)        # ['hi', 'bye', 'hi', 'bye']
print([1, 2] + [3, 4])          # + joins two lists
````

## 1.2 List methods
| Method | What it does |
|---|---|
| `append(x)` | add `x` at the end |
| `insert(i, x)` | add `x` at position `i` |
| `extend(list2)` | add **all items** of another list |
| `remove(x)` | remove the **first** `x` |
| `pop(i)` | remove and **return** item at `i` (last if no `i`) |
| `index(x)` 🆕 | position of the first `x` |
| `count(x)` | how many times `x` appears |
| `sort()` / `reverse()` | change the list **in place** |
| `clear()` 🆕 | remove everything |

````python
nums = [5, 3, 8]

nums.append(1)          # [5, 3, 8, 1]
nums.insert(1, 99)      # [5, 99, 3, 8, 1]
nums.extend([7, 7])     # [5, 99, 3, 8, 1, 7, 7]
print(nums)

nums.remove(99)         # removes the value 99
last = nums.pop()       # removes and returns the last item
print("popped:", last, "->", nums)

print(nums.index(8))    # 🆕 8 is at position 2
print(nums.count(7))    # 7 appears 1 time now

nums.sort()             # changes nums itself
print(nums)
nums.reverse()
print(nums)

print(sorted(nums))     # sorted() returns a NEW list, original stays the same
print(nums)

nums.clear()            # 🆕
print(nums)             # []
````

📝 **Exam tip:** `append` adds ONE item (even a whole list as a single item). `extend` adds each item separately.

````python
a = [1, 2]
a.append([3, 4])
print(a)            # [1, 2, [3, 4]]

b = [1, 2]
b.extend([3, 4])
print(b)            # [1, 2, 3, 4]
````

## 1.3 Copying a list: reference, shallow copy, deep copy 🆕 NEW

🎓 **Teacher tip:** Use this story. *A list is like a Google Doc. Sharing the link (`b = a`) means both of you edit the same document. Making a copy gives you your own document.*

| What we write | What happens |
|---|---|
| `b = a` | **No copy.** `a` and `b` are two names for the same list |
| `b = a.copy()` or `a[:]` or `list(a)` | **Shallow copy**: a new outer list, but inner objects are still shared |
| `b = copy.deepcopy(a)` | **Deep copy**: everything is copied, fully independent |

````python
# 1) Reference assignment: NOT a copy
a = [1, 2, 3]
b = a
b.append(4)
print("a:", a, "| b:", b, "| same object?", a is b)    # both changed!

# 2) Shallow copy: independent for simple lists
c = a.copy()
c.append(5)
print("a:", a, "| c:", c)                              # a is safe

# 3) The shallow-copy problem: nested lists are still shared
nested = [[1, 2], [3, 4]]
shallow = nested.copy()
shallow[0].append(99)
print("nested:", nested)        # [[1, 2, 99], [3, 4]]  <- the original changed too!

# 4) Deep copy solves it
import copy
deep = copy.deepcopy(nested)
deep[1].append(77)
print("nested:", nested)        # unchanged
print("deep  :", deep)
````

📝 **Exam tip:** Use **deep copy** whenever the list contains other lists (or any mutable objects).

## 1.4 List comprehension
**A short way to build a list in one line.**

`[expression  for item in iterable  if condition]`

| Form | Pattern |
|---|---|
| Simple | `[x*2 for x in nums]` |
| Single `if` (filter) | `[x for x in nums if x > 5]` |
| 🆕 Multiple `if` | `[x for x in nums if cond1 if cond2]` (both must be true) |
| 🆕 `if-else` | `[a if cond else b for x in nums]` ← the if-else goes **before** `for` |

````python
nums = list(range(1, 11))

print([x ** 2 for x in nums])                   # squares
print([x for x in nums if x % 2 == 0])          # only even

# 🆕 Multiple if conditions: numbers divisible by 2 AND by 3
print([x for x in range(1, 31) if x % 2 == 0 if x % 3 == 0])     # [6, 12, 18, 24, 30]
# (same as using "and")
print([x for x in range(1, 31) if x % 2 == 0 and x % 3 == 0])

# 🆕 if-else inside the comprehension: goes at the FRONT
print(["even" if x % 2 == 0 else "odd" for x in range(1, 6)])

# Nested: flatten a matrix
matrix = [[1, 2], [3, 4], [5, 6]]
print([item for row in matrix for item in row])     # [1, 2, 3, 4, 5, 6]
````

# Part 2 — Tuples
**A tuple is an ordered but UNCHANGEABLE (immutable) collection**, written with `( )`.

- Faster and uses less memory than a list
- Can be used as a dictionary key (a list cannot)
- A single-item tuple needs a **comma**: `(5,)`

````python
t = (10, 20, 30, 20)
print(t[0], t[-1], t[1:3])
print(t.count(20), t.index(30))

print(type((5)))      # <class 'int'>   -> just brackets around 5, NOT a tuple
print(type((5,)))     # <class 'tuple'> -> the comma makes the tuple

print((1, 2) + (3,))  # join -> (1, 2, 3)
print(("a",) * 3)     # repeat -> ('a', 'a', 'a')
````

## 2.1 Packing and Unpacking 🆕 NEW
- **Packing**: putting many values into one tuple
- **Unpacking**: taking the values out into separate variables (the count must match!)
- `_` is the **disposable variable**: a name we use for a value we do not need
- `*rest` collects "all the remaining items" into a list

````python
# Packing
packed = 1, 2, 3
print(packed)

# Unpacking
a, b, c = packed
print(a, b, c)

# Swap two variables with no temp variable
a, b = b, a
print(a, b)

# Disposable variable: we only want the name and city
name, _, city = ("Asha", 17, "Delhi")
print(name, city)

# Star: collect the rest
first, *rest = (1, 2, 3, 4)
print(first, rest)          # 1 [2, 3, 4]

# Copying a tuple: no need, because it can never change
t = (1, 2, 3)
t2 = t[:]
print(t2 == t)
````

## 2.2 List vs Tuple 🆕 NEW (exam favourite)
| | **List** | **Tuple** |
|---|---|---|
| Brackets | `[ ]` | `( )` |
| Mutable? | Yes | **No** |
| Speed / memory | slower, more memory | faster, less memory |
| Methods | many (`append`, `remove`, ...) | only `count`, `index` |
| Use when | data will change | data must stay fixed (days of week, coordinates) |
| Dictionary key? | No | Yes |

# Part 3 — Sets
**A set is an UNORDERED collection of UNIQUE items**, written with `{ }`.
- No duplicates (extra copies vanish)
- No index (cannot do `s[0]`)
- ⚠️ `{}` makes an empty **dictionary**. For an empty set use `set()`.

````python
s = {1, 2, 2, 3, 3, 3}
print(s)                          # {1, 2, 3}

print(type({}), type(set()))      # dict  set   <- the empty-brackets trap

s.add(4)                          # add one item
s.discard(10)                     # remove if present, NO error if missing
s.remove(1)                       # remove, but ERROR if missing
print(s)

try:
    s.remove(100)
except KeyError:
    print("remove() gives KeyError when the item is not there")
````

## 3.1 Set operations
| Operation | Symbol | Method | Meaning |
|---|---|---|---|
| Union | `a \| b` | `a.union(b)` | everything from both |
| Intersection | `a & b` | `a.intersection(b)` | only common items |
| Difference | `a - b` | `a.difference(b)` | in `a` but not in `b` |
| Symmetric difference | `a ^ b` | `a.symmetric_difference(b)` | in one but not both |

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("union       :", a | b)     # {1, 2, 3, 4, 5, 6}
print("intersection:", a & b)     # {3, 4}
print("difference  :", a - b)     # {1, 2}
print("symmetric   :", a ^ b)     # {1, 2, 5, 6}

# Subset and superset
small = {1, 2}
print(small.issubset(a))          # True  -> every item of small is in a
print(a.issuperset(small))        # True
print(a.isdisjoint({10, 11}))     # True  -> nothing in common

# Frozen set = an immutable set
fs = frozenset([1, 2, 3])
try:
    fs.add(4)
except AttributeError as error:
    print("Error:", error)
````

## 3.2 Set comprehension 🆕 NEW
Same idea as list comprehension, but with `{ }`. Duplicates are removed automatically.

````python
print({x ** 2 for x in range(-3, 4)})                       # squares: -3 and 3 both give 9 -> only one 9
print({x for x in range(20) if x % 2 == 0 if x % 3 == 0})   # multiple if
print({"even" if x % 2 == 0 else "odd" for x in range(10)}) # if-else
````

## 3.3 Performance: why use a set? 🆕 NEW
Checking `x in list` goes through items **one by one** (slow for big data).
Checking `x in set` uses **hashing** (like jumping straight to the right page of a dictionary). It is **much faster**.

````python
import time

big_list = list(range(200_000))
big_set = set(big_list)
target = 199_999            # worst case for the list: the very last item

start = time.perf_counter()
for _ in range(100):
    target in big_list
list_time = time.perf_counter() - start

start = time.perf_counter()
for _ in range(100):
    target in big_set
set_time = time.perf_counter() - start

print(f"list search: {list_time:.5f} sec")
print(f"set  search: {set_time:.5f} sec")
print(f"The set was about {list_time / max(set_time, 1e-9):,.0f} times faster")
# Exact numbers change on every computer. The set will always win.
````

🎓 **Exam tip:** *"Which is faster to search, a list or a set?"* Answer: **set**. Reason: hashing.

# Part 4 — Dictionaries
**A dictionary stores `key: value` pairs**, written with `{ }`. Keys must be **unique and immutable**.

````python
student = {"name": "Asha", "age": 17, "city": "Delhi"}

print(student["name"])              # access with key
print(student.get("marks"))         # None -> .get() never crashes
print(student.get("marks", 0))      # 0    -> default value if key missing

# Add and modify
student["marks"] = 88               # new key -> added
student["age"] = 18                 # existing key -> modified
print(student)

# Delete
print(student.pop("city"))          # removes the key and returns its value
print(student.popitem())            # removes the LAST inserted pair
del student["age"]                  # removes the key
print(student)
student.clear()                     # removes everything
print(student)
````

## 4.1 Looping over a dictionary

````python
marks = {"maths": 90, "science": 85, "english": 78}

for key in marks:                       # by default we get keys
    print(key)

print(list(marks.keys()))               # all keys
print(list(marks.values()))             # all values

for subject, score in marks.items():    # best way: both together
    print(subject, "->", score)
````

## 4.2 Merging dictionaries 🆕 NEW
If a key is in both, the **second dictionary wins**.

````python
d1 = {"a": 1, "b": 2}
d2 = {"b": 20, "c": 30}

# Way 1: update() changes d1 itself
d3 = d1.copy()
d3.update(d2)
print(d3)                  # {'a': 1, 'b': 20, 'c': 30}

# Way 2: ** unpacking operator creates a new dictionary
print({**d1, **d2})

# Way 3: | operator (Python 3.9 and above)
print(d1 | d2)
````

## 4.3 Dictionary comprehension
`{key: value for item in iterable if condition}`

````python
print({x: x ** 2 for x in range(1, 6)})                            # squares

d = {"a": 1, "b": 2, "c": 3, "d": 4}
print({k: v for k, v in d.items() if v > 2})                       # single if

# 🆕 multiple if
print({k: v for k, v in d.items() if v > 1 if v < 4})

# 🆕 if-else (goes in the VALUE part)
print({x: ("even" if x % 2 == 0 else "odd") for x in range(1, 6)})
````

### Classic exam program: count how many times each word appears

````python
sentence = "to be or not to be"
count = {}
for word in sentence.split():
    count[word] = count.get(word, 0) + 1      # .get() with default 0 is the trick
print(count)      # {'to': 2, 'be': 2, 'or': 1, 'not': 1}
````

# Part 6 — Strings
**A string is an ordered, immutable sequence of characters.** Indexing and slicing work exactly like lists.

## 6.1 How a string is stored 🆕 NEW (internal implementation)
- Internally a string is a **sequence of Unicode characters**, and each character has a position (index)
- Each character has a number code: `ord("A")` = 65, and `chr(65)` = "A"
- Strings are **immutable**: every "change" creates a **new** string in memory

````python
s = "Python Revision"
print(s[0], s[-1], s[0:6], s[::-1], len(s))

print(ord("A"), chr(65))        # character <-> number code

try:
    s[0] = "J"                  # cannot change one character
except TypeError as error:
    print("Error:", error)

new_s = "J" + s[1:]             # build a NEW string instead
print(new_s)
````

## 6.2 Important string methods

````python
text = "  Hello, Python World  "

print(text.strip())                       # remove spaces from both ends
print(text.upper(), "|", text.lower())
print(text.strip().title())               # Capital Letter Of Each Word
print(text.replace("Python", "Java"))
print(text.find("Python"))                # position, or -1 if not found
print(text.count("o"))
print(text.strip().startswith("Hello"), text.strip().endswith("x"))
print("123".isdigit(), "abc".isalpha())

words = "one,two,three".split(",")       # string -> list
print(words)
print("-".join(words))                    # list -> string
````

## 6.3 String formatting — quick recap

````python
name, score = "Asha", 91.5
print("Name: %s, Score: %.1f" % (name, score))      # % style
print("Name: {}, Score: {:.1f}".format(name, score)) # format()
print(f"Name: {name}, Score: {score:.1f}")           # f-string (easiest)
print(", ".join(["red", "green", "blue"]))           # join()

# Palindrome check
word = "madam"
print(word, "is palindrome" if word == word[::-1] else "is not palindrome")
````

# Part 7 — Regular Expressions
**A regex is a pattern used to search, extract, validate or replace text.** We use Python's `re` module. Always write patterns as **raw strings**: `r"..."`.

| Function | Use |
|---|---|
| `re.match(p, s)` | match only at the **start** of the string |
| `re.search(p, s)` | find the **first** match anywhere |
| `re.findall(p, s)` | list of **all** matches |
| `re.sub(p, new, s)` | **replace** matches |
| `re.split(p, s)` | **split** at matches |
| `re.compile(p)` | save a pattern to reuse |

````python
import re

text = "Order 123 shipped, order 456 pending"

print(re.match(r"Order", text))          # matches at the start -> match object
print(re.match(r"\d+", text))            # None, text does not START with digits
print(re.search(r"\d+", text).group())   # '123' (first match)
print(re.findall(r"\d+", text))          # ['123', '456']

print(re.sub(r"\d+", "#", text))         # replace numbers with #
print(re.split(r"[,;]\s*", "a, b;c,   d"))  # split on comma or semicolon

pattern = re.compile(r"\d{3}-\d{4}")     # compile once, use many times
print(pattern.findall("555-1234 and 777-9999"))
````

## 7.1 Quantifiers and character classes
| Quantifier | Meaning | | Class | Meaning |
|---|---|---|---|---|
| `*` | 0 or more | | `\d` | digit |
| `+` | 1 or more | | `\w` | letter, digit or `_` |
| `?` | 0 or 1 | | `\s` | whitespace |
| `{n}` | exactly n | | `.` | any character |
| `{n,m}` | n to m | | `[abc]` | a, b or c |
| `^` start · `$` end | | | `[^abc]` | anything except a, b, c |

**Flags:** `re.I` (ignore case), `re.M` (^ and $ per line), `re.S` (dot matches newline)

````python
print(re.findall(r"ab*", "a ab abb"))              # ['a', 'ab', 'abb']
print(re.findall(r"\d{2,3}", "1 12 123 1234"))     # ['12', '123', '123']
print(re.findall(r"[aeiou]", "regular expressions"))   # vowels
print(re.findall(r"python", "Python PYTHON", re.I))    # ignore case flag

# Greedy vs lazy (adding ? makes it lazy)
html = "<b>bold</b><i>italic</i>"
print(re.findall(r"<.+>", html))      # greedy: one big match
print(re.findall(r"<.+?>", html))     # lazy: each tag separately

# Groups with ( )
m = re.search(r"(\w+)@(\w+)\.com", "mail: sam@gmail.com")
print(m.group(), m.group(1), m.group(2))

# Practical: extract email addresses
msg = "Contact info@example.com or sales@shop.co.in"
print(re.findall(r"[\w.+-]+@[\w-]+(?:\.[\w-]+)+", msg))

# Practical: valid 6-digit PIN code?
print(bool(re.fullmatch(r"\d{6}", "226001")), bool(re.fullmatch(r"\d{6}", "22600")))
````

# Part 8 — Exception Handling
**An exception is an error that happens while the program is running.** Without handling, the program crashes. With `try/except`, we handle it politely.

| Block | Runs when |
|---|---|
| `try` | always first: put the risky code here |
| `except` | only if an error happens in `try` |
| `else` | only if **no** error happened |
| `finally` | **always**, error or not (used for clean-up) |

````python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("You cannot divide by zero")
    except TypeError:
        print("Please give numbers only")
    else:
        print("Division worked:", result)
    finally:
        print("-- finished --")

safe_divide(10, 2)
safe_divide(10, 0)
safe_divide(10, "a")
````

## 8.1 Grouping multiple exceptions 🆕 NEW
Put the exceptions in a **tuple** to handle them with the same code. `as e` gives us the error message.

````python
for value in ["10", "abc", None]:
    try:
        print(100 / int(value))
    except (ValueError, TypeError) as e:       # one except for two errors
        print("Bad input:", repr(value), "->", e)
````

## 8.2 Raising and re-raising exceptions 🆕 NEW
- `raise` creates an error **on purpose**
- A bare `raise` inside `except` **re-throws** the same error upward after we have done something (like logging)

````python
# Raise on purpose
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    return age

try:
    set_age(-5)
except ValueError as e:
    print("Caught:", e)

# Re-raise: handle a little, then pass it up
try:
    try:
        1 / 0
    except ZeroDivisionError:
        print("Inner: logging the error...")
        raise                                    # throw the same error again
except ZeroDivisionError:
    print("Outer: caught the re-raised error")
````

## 8.3 Custom exceptions 🆕 NEW
We can create our own error type. 🎓 *`class` will be taught in OOP, so tell students: "just copy this pattern for now".*

````python
class NegativeAgeError(Exception):      # our own error type
    pass

def check_age(age):
    if age < 0:
        raise NegativeAgeError("Age cannot be negative")
    return age

try:
    check_age(-3)
except NegativeAgeError as e:
    print("Custom error caught:", e)
````

## 8.4 Exception hierarchy 🆕 NEW
Exceptions form a **family tree**. Catching a parent also catches all its children.

```
BaseException
 └── Exception
      ├── ArithmeticError
      │     ├── ZeroDivisionError
      │     └── OverflowError
      ├── LookupError
      │     ├── IndexError
      │     └── KeyError
      ├── ValueError
      ├── TypeError
      ├── NameError
      ├── OSError
      │     └── FileNotFoundError
      └── (your own custom exceptions)
```

````python
# Python can show the family line (child -> parent -> ... -> object)
print([c.__name__ for c in ZeroDivisionError.__mro__])
print([c.__name__ for c in KeyError.__mro__])

# Catching the parent catches the child
try:
    {}["missing"]
except LookupError:                     # KeyError is a child of LookupError
    print("Caught by the parent: LookupError")
````

# Part 9 — File Handling

| Mode | Meaning |
|---|---|
| `"r"` | **read** (default). Error if the file does not exist |
| `"w"` | **write**. Creates the file, **erases old content** if it exists |
| `"a"` | **append**. Adds at the end, keeps old content |
| `"x"` | **create only**. Error if the file already exists |

**Best practice: use `with open(...)`**. It closes the file automatically, even if an error happens.

````python
# Write (creates demo.txt)
with open("demo.txt", "w", encoding="utf-8") as f:
    f.write("Line 1\n")
    f.write("Line 2\n")

# Append (adds to the end)
with open("demo.txt", "a", encoding="utf-8") as f:
    f.write("Line 3\n")

# Read everything
with open("demo.txt", "r", encoding="utf-8") as f:
    print(f.read())
````

````python
with open("demo.txt", "r", encoding="utf-8") as f:
    print(repr(f.readline()))      # one line (includes the \n)
    print(f.readlines())           # the remaining lines as a list
    print("-- cursor position:", f.tell())   # tell() = where the cursor is (in bytes)
    f.seek(0)                      # seek(0) = go back to the start
    for line in f:                 # looping over the file reads line by line
        print(line.strip())

# "x" mode fails if the file exists
try:
    with open("demo.txt", "x") as f:
        pass
except FileExistsError:
    print("x mode: demo.txt already exists, so Python refused")
````

## 9.1 File-to-file transfer 🆕 NEW
Read from one file and write into another (a copy program).

````python
with open("demo.txt", "r", encoding="utf-8") as source, \
     open("copy.txt", "w", encoding="utf-8") as target:
    for line in source:
        target.write(line.upper())      # we can change the text while copying

with open("copy.txt", "r", encoding="utf-8") as f:
    print(f.read())

# Clean up the files we made
import os
os.remove("demo.txt")
os.remove("copy.txt")
print("files exist now?", os.path.exists("demo.txt"))
````

# Part 10 — Modules and Packages
- **Module** = one Python file (`.py`) with functions or variables that we can import
- **Package** = a **folder** of modules that has a file named `__init__.py`

````python
import math                       # import the whole module
print(math.sqrt(16), math.pi)

import math as m                  # import with a short name
print(m.floor(3.7))

from math import pi, sqrt         # import only what we need (no "math." needed)
print(sqrt(25), pi)

# ---- Make our OWN module: any file saved as .py is a module ----
with open("my_tools.py", "w") as f:
    f.write("def add(a, b):\n    return a + b\n")

import my_tools
print(my_tools.add(2, 3))

# ---- Make our OWN package: a folder + __init__.py ----
import os
os.makedirs("shapes_pkg", exist_ok=True)
open("shapes_pkg/__init__.py", "w").close()           # empty file marks it as a package
with open("shapes_pkg/area.py", "w") as f:
    f.write("def square(side):\n    return side * side\n")

from shapes_pkg import area
print(area.square(4))

# Clean up
import shutil
os.remove("my_tools.py")
shutil.rmtree("shapes_pkg")
shutil.rmtree("__pycache__", ignore_errors=True)
````
